Ingest Files into Bronze Tables

```python
INGESTION_CONFIG = [
    {
        "source": "crm",
        "path": "/Volumes/workspace/bronze/raw_sources/cust_info.csv",
        "table": "crm_cust_info"
    }
]
```

```python
for item in INGESTION_CONFIG:

    print(f"Ingesting {item['source']} → workspace.bronze.{item['table']}")

    df = (
        spark.read
            .option("header", "true")
            .option("inferSchema", "true")
            .csv(item["path"])
    )

    (
        df.write
            .mode("overwrite")
            .option("overwriteSchema", "true")
            .format("delta")
            .saveAsTable(f"workspace.bronze.{item['table']}")
    )

    print(f"Successfully ingested {item['table']}")
```
